```scala
def grouped[A](l: List[A], size: Int): List[List[A]] = 
    l match 
        case Nil => List()
        case head :: tail => 
            if size == 1 then 
                List(head) :: grouped(tail, size)
            else 
                grouped(tail, size-1) match 
                    case head2 :: tail2 => 
                        (head :: head2) :: grouped(tail2.flatten, size)
                    case Nil => 
                        List(head) :: Nil
```
