```scala
def pascal(n: Int): IndexedSeq[IndexedSeq[Int]] = {
    
    def newrow(maybeRow: Option[IndexedSeq[Int]]): IndexedSeq[Int] = 
        maybeRow.fold(IndexedSeq(1))(row => 
            (1 +: 
             (if (row.length == 1) IndexedSeq() 
              else row.sliding(2).map{ case IndexedSeq(a: Int, b: Int) => a+b }.toIndexedSeq)) :+ 1)


    if (n <= 0) 
        IndexedSeq()
    else {
        val solaux = pascal(n-1)
        solaux :+ newrow(solaux.lastOption)
    }   
}
```

```scala
pascal(10).foreach(println(_))
```
