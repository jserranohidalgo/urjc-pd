```python
def containsSuch(dict, key, pred): 
    return key in dict and pred(dict[key])
    
def modifyOrAdd(dict, k, value, upd):
    if k in dict:
        upd(dict[k])
    else: 
        dict[k] = value
        
def modify(dict, k, upd): 
    if k in dict: 
        upd(dict[k])
    else: 
        dict
```

```python
def tag(word, guess): 
    tagged = []
    index = 0
    maybeYellow = {}
    occurrences = {}
    
    for (c, g) in zip(list(word), list(guess)):
        matched = c == g
        matchedCurrent = containsSuch(occurrences, c, lambda v: v > 0)
        matchedGuess = containsSuch(maybeYellow, g, lambda v: v != [])
        
        if matched: 
            tagged.append((c, 'green'))
            index += 1 
        elif not matched and not matchedCurrent and not matchedGuess:
            modifyOrAdd(maybeYellow, c, [index], lambda l: l.append(index))
            modifyOrAdd(occurrences, g, 1, lambda o: o + 1)
            tagged.append((c, "Gray"))
            index += 1
        elif not matched and not matchedCurrent and matchedGuess:
            modifyOrAdd(maybeYellow, c, [index], lambda l: l.append(index))
            modify(maybeYellow, g, lambda h, *tail:  tail)
            tagged[maybeYellow[g][0]] = (g, "Yellow")
            tagged.append((c, "Gray"))
            index += 1
        elif not matched and matchedCurrent and not matchedGuess:
            modify(occurrences, c, lambda o: o - 1)
            tagged.append((c, "Yellow"))
            index += 1
        else:
            modify(maybeYellow, g, lambda h, *tail: tail)
            modify(occurrences, c, lambda o: o - 1)
            tagged[maybeYellow[g][0]] = (g, "Yellow")
            tagged.append((c, "Yellow"))
            index += 1
            
    return tagged
```

```python
print(tag("hlioe", "hello"))
print(tag("hille", "helli"))
```
