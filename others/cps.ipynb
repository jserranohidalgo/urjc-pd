### Excluded third from double negation

```scala
def EM_Irrefutable[A]: Not[Not[A Either Not[A]]] = 
    (nem: A Either Not[A] => Nothing) => nem(Right((a: A) => nem(Left(a)) ) : A Either Not[A]) : Nothing
```

```scala
type DN = { def apply[P]: Not[Not[P]] => P }
type EM = { def apply[P]: P Either Not[P] }
```

```scala
def proof(dn: DN): EM = new {
    def apply[P]: P Either Not[P] = 
        dn[P Either Not[P]](EM_Irrefutable[P])
}
```

## continuations as a programming abstraction

```scala
def multiply0(l: List[Int]): Int = 
    l match {
        case Nil => 1
        case head :: tail => 
            head * multiply(tail)
    }
```

```scala
def multiply1(l: List[Int]): Int = 
    l match {
        case Nil => 1
        case 0 :: _ => 0
        case head :: tail => 
            head * multiply(tail)
    }
```

```scala
def multiply2[A](l: List[Int])(cont: (Int => A)): A = 
    l match {
        case Nil => cont(1)
        case 0 :: _ => cont(0)
        case head :: tail => 
            multiply2(tail)(multTail => cont(multTail*head))
    }
```
