```scala
trait Functor[F[_]]{
    def map[A, B](fa: F[A])(f: A => B): F[B] 
}

object Functor{
    def apply[F[_]](implicit F: Functor[F]) = F
}
```

```scala
trait Isomorphic[A, B]{
    
    def from(a: A): B
    
    def to(b: B): A
    
    // equality 
    
    def equalA(a1: A, a2: A): Boolean = 
        a1 == a2
    
    def equalB(b1: B, b2: B): Boolean =
        b1 == b2
    
    // Bijection laws
    
    def law1(a: A): Boolean = 
        equalA(to(from(a)), a)
    
    def law2(b: B): Boolean = 
        equalB(from(to(b)), b)
}
```

```scala
class Iso[F[_]: Functor, A] 
extends Isomorphic[[B] => (A => B) => F[B], F[A]]
{
    def from(f: [B] => (A => B) => F[B]): F[A] = 
        f(a => a)
    
    def to(fa: F[A]): [B] => (A => B) => F[B] = 
        [b] => (f: A => b) => Functor[F].map(fa)(f)
}
```

### Wadler: a taste of liner logic

```scala
def `x-E`[A, B, C](t: (A, B), f: A => B => C): C = 
    f(t._1)(t._2)
```

```scala
def `x-I`[A, B](a: A, b: B): (A, B) = 
    (a, b)
```

```scala
def `x-E'`[A, B](t: (A, B)): A = 
    `x-E`(t, (a, b) => a)
```
