```python
def square(n): 
    output = []
    for i in range(n):
        output.append((i, i*i))
    return output
```

```python
print(square(0))
print(square(1))
print(square(2))
print(square(3))
print(square(4))
print(square(5))
print(square(6))

```

```python
def square(n): 
    return n*n
```

```python
def inc1(n): 
    return n+1
```

```python
def double(n): 
    return 2*n
```

```python
def first(n, f): 
    output = []
    for i in range(n):
        output.append((i, f(i)))
    return output
```

```python
first(5, double)
```

```python
first(5, inc1)
```

```python
first(5, square)
```

```python
from random import randint
import random

N = random.randint(1,4)
LP = ['a', 'RELOJ',  'GLOBO', 'VIAJE',  'AMIGO', ]
PAA = LP[N]
print(PAA)
LPAA = len(PAA)
o = -1
list(PAA)

for i in range(5):
    PI = input(f"Escriba una palabra de {LPAA} letras, en mayúsculas: ")
    list(PI)

for i in range(5):
    o += 1
    v = []
    v.append(PI[o])
    b = " ".join(v)
    pl = b in PAA
    if pl == True:
        ppu = PI.index(b)
        ppa = PAA.index(b)
        if ppa == ppu:
            print(f"{v}(verde)")
        if ppa != ppu:
            print(f"{v}(amarillo)")
    else:
        print(f"{v}(gris)")

    o = -1
```
